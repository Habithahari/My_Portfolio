```python
# import libs
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('land_prices.csv')
print(df.head())
```

```python
df["price_per_cent"] = df['price_lakhs']/df['land_area_cents']
print(df.head())
```

```python
print(df.columns)
```

```python
from pandas.core.tools import numeric
numeric_features = ['land_area_cents', 'distance_to_school_km',
       'distance_to_airport_km', 'distance_to_railway_station_km',
       'distance_to_hospital_km', 'distance_to_medical_college_km',
       'distance_to_bus_stop_km', 'distance_to_market_km']
#convert text to numeric
#one-hot encoding
df_processed = pd.get_dummies(df, columns=['location_name', 'taluk', 'village','land_type'], drop_first=True)
print(df_processed.head())
```

```python
dummy_features = []
for column_name in df_processed.columns:
  if column_name.startswith('land_type') or column_name.startswith('location_name') or column_name.startswith('taluk') or column_name.startswith('village'):
    dummy_features.append(column_name)
print(dummy_features)
```

```python
features = numeric_features + dummy_features
print(features)
```

```python
# x = features
# y = price per cent
x = df_processed[features]
y = df_processed['price_per_cent']

# split dataset 80/20

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# choose, create and train model

from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
```

```python
#evaluation
y_pred = model.predict(x_test)
df_comparison = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})
print(df_comparison.head())
```

```python

# actual vs predicted scatterplot

import plotly.express as px
import pandas as pd

results = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': y_pred
})

fig = px.scatter(
    results,
    x='Actual Price',
    y='Predicted Price',
    title='Actual vs Predicted Land Prices',
    hover_data=['Actual Price', 'Predicted Price']
)

# Ideal prediction line
min_val = min(results.min())
max_val = max(results.max())

fig.add_shape(
    type='line',
    x0=min_val, y0=min_val,
    x1=max_val, y1=max_val,
    line=dict(dash='dash', color='red')
)

fig.update_layout(
    template='plotly_white',
    height=550
)

fig.show()
```

```python

fig = px.histogram(
    df,
    x='price_per_cent',
    nbins=30,
    title='Distribution of Land Prices Per Cent',
    color_discrete_sequence=['#3498db']
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Price Per Cent',
    yaxis_title='Number of Land Records'
)

fig.show()
```

```python

location_prices = (
    df.groupby('location_name')['price_per_cent']
    .mean()
    .reset_index()
    .sort_values('price_per_cent', ascending=False)
)

fig = px.bar(
    location_prices,
    x='location_name',
    y='price_per_cent',
    title='Average Land Price by Location',
    color='price_per_cent',
    color_continuous_scale='Viridis'
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Location',
    yaxis_title='Average Price Per Cent',
    xaxis_tickangle=-45
)

fig.show()
```
